# What is a CSV File?
---
* **CSV** stands for **Comma-Separated Values**
* Each line represents a row; commas separate the cells
* Lacks Excel features — no types, no formatting, no charts
* But that simplicity is also its strength — universally readable

```
name,department,salary
Alice,Engineering,85000
Bob,Marketing,72000
```

* Each line in a CSV file represents a row in a spreadsheet
* The commas separate the cells in the row
* The advantage of CSV files is that they are simple
* CSV is widely supported by many types of programs
* CSV files are simple, and lack many of the features of an Excel spreadsheet, like:
	* They don't have types — everything is a string
	* No settings for fonts, size, or color
	* Do not have multiple worksheets
	* No merged cells
	* No images or charts
* Since CSV files are just text you can read them in as a string
* As each cell is separated by a comma you could split the cells on the commas
* The issue is not every comma may represent a boundary between cells
* CSV has escape characters that allow commas to be part of values, which `split()` would not properly handle
* As a result using the `csv` module is the better choice

# Reading CSV Files
---
* The `csv` module comes with Python — no installation needed
* Open the file with `open()`, pass to `csv.reader()` to get a **Reader** object
* Convert to a list with `list()` to get a **list of lists**
* Access individual values with `data[row][col]`

* To read data from a CSV file with the `csv` module a `Reader` object is needed
* A `Reader` object allows you to iterate over the lines in a CSV file
* The `csv` module comes with Python, so it doesn't need to be installed through pip
* To work with a CSV file the `csv` module needs to be imported
* To read a CSV file first it needs to be opened with `open()`
* Then the file object `open()` returns will be passed to `csv.reader()`
* `csv.reader()` will return a `Reader` object for you to utilize
* The most direct way to access values in a `Reader` object is to pass it to `list()`
* Using `list()` will return a list of lists
* This list of lists can then be stored in a variable
* The data can be printed to show it is a list of lists

````python
import csv

# Open the file and create a Reader object
# (assumes example.csv exists in the working directory)
with open('example.csv') as csv_file:
    csv_reader = csv.reader(csv_file)
    data = list(csv_reader)  # convert to list of lists

print(data)
print()

# Access individual values with [row][col]
print(data[0])     # first row (often headers)
print(data[1][0])  # first value of second row
````

# Working With Headers
---
* The first row of a CSV is often a **header row** — column labels, not data
* Pull it out separately so you can work with just the data rows
* Use `next()` on the reader to grab the first row before looping

````python
import csv

with open('example.csv') as csv_file:
    csv_reader = csv.reader(csv_file)

    # Pull out the header row first
    headers = next(csv_reader)
    print(f"Headers: {headers}")

    # Now loop through only the data rows
    for row in csv_reader:
        print(row)
````

# Looping Through a Reader Object
---
* For larger files, loop through the Reader directly — avoids loading everything into memory
* The Reader object can only be looped **once** — call `csv.reader()` again to re-read
* Each row comes back as a **list of strings**

* For larger CSV files, it may be best to use a `Reader` object in a `for` loop
* This avoids loading the entire file into memory at once
* This can be done once you have the reader object
* You can loop through the rows in the reader object, much like you would a list
* The reader object can be looped over only once
	* To re-read the CSV file you must call `csv.reader()` to create a new `Reader`

````python
import csv

with open('example.csv') as csv_file:
    csv_reader = csv.reader(csv_file)
    headers = next(csv_reader)

    # Loop through row by row — efficient for large files
    for row_num, row in enumerate(csv_reader, start=1):
        print(f"Row {row_num}: {row}")
````

# Writing CSV Files
---
* A **Writer** object lets you write data to a CSV file
* Open with `'w'` mode, pass to `csv.writer()`
* Use `.writerow()` with a list — each item becomes a cell
* The writer automatically escapes any commas in values

* A `Writer` object lets you write data to a CSV file
* To create a `Writer` object the `csv.writer()` function is called
* Again `open()` needs to be called first, but it needs an additional argument
	* You need to pass `'w'` to indicate the file will be opened for writing
* Then the `writerow()` method for `Writer` objects can be used
	* It takes a list of arguments
	* Each value in the list is placed in its own cell
* The writer will automatically escape any commas in the values

````python
import csv

# Write data to a CSV file
with open('output.csv', 'w', newline='') as csv_file:
    csv_writer = csv.writer(csv_file)

    # Write the header row
    csv_writer.writerow(['Name', 'State', 'Population'])

    # Write data rows
    csv_writer.writerow(['Laramie', 'Wyoming', 32000])
    csv_writer.writerow(['Cheyenne', 'Wyoming', 65000])
    csv_writer.writerow(['Casper', 'Wyoming', 58000])

# Verify by reading it back
with open('output.csv') as csv_file:
    print(csv_file.read())
````

# Custom Delimiters
---
* Sometimes data uses a different separator — like a **tab** instead of a comma
* The **delimiter** is the character between cells (default: `,`)
* The **line terminator** is the character at the end of each row (default: `\n`)
* Tab-separated files use the `.tsv` extension by convention

* Suppose you want to separate values with something like a tab character rather than commas
* Or, you want rows to be double spaced
* You can change the delimiter and line terminator characters in your file
	* The *delimiter* is the character that appears between cells on a row, by default a comma
	* The *line terminator* is the character that comes at the end of a row, by default a new line
* Passing `delimiter='\t'` changes the character between cells to be a tab character
* Passing `lineterminator='\n\n'` changes the character between rows to be two newline characters
* The file being written to ended with `.tsv` to denote it is a *tab separated values* file

````python
import csv

# Write a tab-separated file instead of comma-separated
with open('output.tsv', 'w', newline='') as tsv_file:
    tsv_writer = csv.writer(tsv_file, delimiter='\t', lineterminator='\n')
    tsv_writer.writerow(['Name', 'State', 'Population'])
    tsv_writer.writerow(['Laramie', 'Wyoming', 32000])
    tsv_writer.writerow(['Cheyenne', 'Wyoming', 65000])

# Verify
with open('output.tsv') as tsv_file:
    print(tsv_file.read())
````
