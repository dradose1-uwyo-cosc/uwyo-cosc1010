# What is JSON?
---
* **JSON** = JavaScript Object Notation — despite the name, you don't need to know JavaScript
* A human-readable format for storing and transmitting data as `key:value` pairs
* JSON can only store certain types:
    * strings, integers, floats, Booleans, lists, dictionaries, NoneType
* JSON strings **always use double quotes**

* Python's `json` module handles all the details of translating between a string with JSON data and Python values
* This occurs with the `json.loads()`, to load the JSON data
* And with `json.dumps()` to write the data out
* JSON cannot represent Python-specific objects
* JSON can't store *every* kind of Python value
* It can contain only:
	* strings, integers, floats, Booleans, lists, dictionaries, NoneType

# Reading JSON with `json.loads()`
---
* `loads()` stands for **load string** — converts a JSON string into a Python value
* Import the `json` module first — it comes with Python
* Returns a Python dictionary, list, or other appropriate type
* JSON objects can be nested, just like Python dictionaries

* To translate a string containing JSON into a Python value, the string needs to be passed to `json.loads()`
	* The name means *load string*
* First though, the `json` module must be imported
* Then `json.loads()` can be called
* JSON strings always use double quotes
* It will return the data as a Python dictionary
* Much like we could have dictionaries in dictionaries, you can have JSON objects nested in others

```python
import json

# A JSON string — note the double quotes
json_string = '{"name": "University of Wyoming", "founded": 1886, "city": "Laramie"}'

# Convert JSON string to a Python dictionary
data = json.loads(json_string)
print(type(data))
print(data)
print(data['name'])
print(data['founded'])
```

```python
import json

# Nested JSON — objects within objects
json_string = '''
{
    "university": "University of Wyoming",
    "location": {
        "city": "Laramie",
        "state": "Wyoming",
        "elevation_ft": 7220
    },
    "colleges": ["Engineering", "Arts and Sciences", "Business"]
}
'''

data = json.loads(json_string)
print(data['location']['city'])
print(data['location']['elevation_ft'])
print(data['colleges'][0])
```

# Writing JSON with `json.dumps()`
---
* `dumps()` stands for **dump string** — converts a Python value into a JSON string
* The value must be one of the JSON-compatible types
* Use `indent=` for pretty-printed, human-readable output

* `dumps()` stands for *dump string*
* It translates a Python value into a string of JSON formatted data
* The value can only be one of the following types:
	* dictionaries, lists, integers, floats, strings, Booleans, None

```python
import json

# A Python dictionary
uw = {
    'name': 'University of Wyoming',
    'founded': 1886,
    'city': 'Laramie',
    'colleges': ['Engineering', 'Arts and Sciences', 'Business']
}

# Convert to a JSON string
json_string = json.dumps(uw)
print(type(json_string))
print(json_string)

# Pretty-print with indentation
print()
print(json.dumps(uw, indent=4))
```

# Reading and Writing JSON Files
---
* Combine `pathlib` with `json` to read and write JSON files
* Read: `json.loads(path.read_text())`
* Write: `path.write_text(json.dumps(data, indent=4))`

```python
import json
from pathlib import Path

# Write data to a JSON file
uw = {
    'name': 'University of Wyoming',
    'founded': 1886,
    'city': 'Laramie',
    'colleges': ['Engineering', 'Arts and Sciences', 'Business']
}

path = Path('uw.json')
path.write_text(json.dumps(uw, indent=4))
print("Written to uw.json")

# Read it back
data = json.loads(path.read_text())
print(f"\nRead back: {data['name']}")
print(f"Colleges: {data['colleges']}")
```

# JSON and APIs
---
* Many websites provide data through an **API** (Application Programming Interface)
* APIs return data as JSON — you make a request to a URL and get JSON back
* The **National Weather Service** has a free API that returns forecasts as JSON
* Use the `requests` module to make API calls

* We can put it together to write a program to give us a weather forecast
* We can:
	* Make an API call to retrieve JSON
	* Parse that string into a Python dictionary
	* Give a forecast for Laramie
* To begin we will need an additional module to make the API request
	* API requests are very similar to when you visit a website
* We will be using the `requests` module to accomplish this
* OK, we have the JSON data — the file is a lot
* The data we really want is a sub dictionary of a list of a sub dictionary
* We can read in the overall data, and then using the keys find the data we really want
* Looking at your data ahead of time really helps to know what you have access to

```python
import json
import requests

# National Weather Service API — free, no key needed!
url = 'https://api.weather.gov/gridpoints/CYS/84,23/forecast'

try:
    response = requests.get(url)
    data = json.loads(response.text)

    # Dig into the nested structure to find the forecast periods
    periods = data['properties']['periods']

    # Print the next few forecast periods
    for period in periods[:3]:
        print(f"{period['name']}: {period['detailedForecast']}")
        print()

except Exception as e:
    print(f"Could not retrieve forecast: {e}")
```
