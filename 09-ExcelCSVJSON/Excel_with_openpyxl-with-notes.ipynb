# Excel Terminology
---
Before diving in, some important terms:
* A spreadsheet document is called a **workbook** — saved as a file
* Each workbook can have multiple **sheets**
* The sheet currently being viewed is the **active sheet**
* Sheets have **columns** (letters starting at A) and **rows** (numbers starting at 1)
* A box at a particular row and column is a **cell**

* First to get some basic semantics out of the way:
	* An Excel spreadsheet document is called a *workbook*
	* A single workbook is saved in a *file*
	* Each workbook can have multiple *sheets*
	* The sheet being viewed is the *active sheet*
* Each sheet has:
	* *columns* addressed by letters starting with A
	* *rows* addressed by numbers starting at 1
* A box at a particular row and column is a *cell*
* Each cell can contain a number or text value
* The grid of cells with data make up a sheet

# Working With Sheets
---
* Get a list of all sheet names with the `sheetnames` attribute
* Access individual sheets with `wb['SheetName']` — like a dictionary
* Get the currently active sheet with `wb.active`

* Once you have a workbook you can get a list of all the sheets
* This is done with the `sheetnames` attribute
* Each sheet is represented by a `Worksheet` object
* The individual sheets can be obtained using the `[]` and the sheet name, much like when accessing a value in a dictionary
* The `active` attribute can be used to retrieve the active sheet

```python
import openpyxl

wb = openpyxl.load_workbook('example.xlsx')

# List all sheet names
print(wb.sheetnames)

# Get a specific sheet by name
sheet = wb['Sheet1']
print(type(sheet))

# Get the active sheet
active_sheet = wb.active
print(active_sheet.title)
```

# Accessing Cells
---
* Access cells by name using `sheet['B1']`
* Each `Cell` object has a **value** attribute containing what's stored
* Also has `row`, `column`, and `coordinate` attributes
* Use `.cell(row, column)` to access by numbers — rows and columns start at **1**, not 0

* Once you have a `Worksheet` object cells can be accessed by their name
* The `Cell` object has a `value` attribute that contains the value stored in the cell
* `Cell` objects also have `row`, `column`, and `coordinate` attributes that provide location information
* Notice the attribute `column` gives a number, rather than the expected letter
* Instead `column_letter` needs to be used to get the letter
* Alternatively a cell can be retrieved with the `.cell()` method
* `.cell()` accepts integers for both row and column
	* The first row or column is **1** not 0
* You can also write a `for` loop to go through the values for a series of cells
* You can determine the size of a sheet with two attributes of `Worksheet`:
	* `.max_row`
	* `.max_column` — returns an `int` rather than a letter

```python
import openpyxl

wb = openpyxl.load_workbook('example.xlsx')
sheet = wb.active

# Access by cell name
cell = sheet['A1']
print(f"Value: {cell.value}")
print(f"Row: {cell.row}, Column: {cell.column}")
print(f"Coordinate: {cell.coordinate}")

# Access by row and column numbers (start at 1!)
cell_b1 = sheet.cell(row=1, column=2)
print(f"B1 value: {cell_b1.value}")

# Sheet dimensions
print(f"Max rows: {sheet.max_row}, Max columns: {sheet.max_column}")
```

# Reading Data from a Sheet
---
* Loop through rows and columns to read all data
* Use `max_row` and `max_column` to know when to stop
* Worksheet objects can also be **sliced** to get a range of cells

```python
import openpyxl

wb = openpyxl.load_workbook('example.xlsx')
sheet = wb.active

# Loop through all rows and columns
for row in range(1, sheet.max_row + 1):
    for col in range(1, sheet.max_column + 1):
        cell = sheet.cell(row=row, column=col)
        print(cell.value, end='\t')
    print()  # newline after each row
```

# Creating and Writing to Excel Files
---
* Create a blank workbook with `openpyxl.Workbook()`
* Write to cells just like writing to a dictionary: `sheet['A1'] = value`
* **Always save with `.save(filename)`** — changes are not saved automatically
* When working with existing files, save to a **copy** — never overwrite the original!

* We aren't limited to only reading from an Excel file
* `openpyxl` also provides ways of writing data
* Meaning, your programs can create and edit spreadsheet files
* `openpyxl.Workbook()` can be called to create a new, blank `Workbook` object
* The workbook starts off with a single sheet, named `Sheet`
* The name of the sheet can be changed by storing a new value in the `title` attribute
* Anytime the `Workbook` is modified it will not be saved until `.save()` is called, with a name given
* Writing values to cells is much like writing values to keys in a dictionary
* You supply the cell's coordinates as a string in `[]` for the sheet
* When working with sheets you should always save to a copy, not modify the original when working with a pre-populated data set
* This ensures you have a backup in case you mess up

```python
import openpyxl

# Create a new blank workbook
wb = openpyxl.Workbook()
sheet = wb.active
sheet.title = 'Wyoming Cities'  # rename the default sheet

# Write headers
sheet['A1'] = 'City'
sheet['B1'] = 'Population'
sheet['C1'] = 'County'

# Write data rows
data = [
    ('Cheyenne', 65000, 'Laramie County'),
    ('Casper', 58000, 'Natrona County'),
    ('Laramie', 32000, 'Albany County'),
]
for row_num, (city, pop, county) in enumerate(data, start=2):
    sheet.cell(row=row_num, column=1).value = city
    sheet.cell(row=row_num, column=2).value = pop
    sheet.cell(row=row_num, column=3).value = county

# Save the workbook
wb.save('wyoming_cities.xlsx')
print("Saved wyoming_cities.xlsx")
```
