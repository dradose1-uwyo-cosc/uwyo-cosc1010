# Why this matters
---
Every decision a program makes starts with a question
* Is this user logged in?
* Is the temperature above freezing?
* Is this item in stock?

Before your code can make decisions, it needs to be able to **ask questions and get a yes or no answer** — that's what conditional tests do

* We are going to start off with a test that lets you respond to situations correctly
* Imagine you have a list of cities
* And you want to print each city with proper casing
* City names are proper nouns, so should begin with a capital
* But what if you want only a specific city in all upper?
* That code follows a loop through the list
* It looks for the value `'laramie'`, printing in upper case when found
* All other cases are printed in title case
* This combines many of the things we will talk about

```python
# A preview of what if statements can do
cities = ["laramie", "casper", "cheyenne", "sheridan", "gillette"]

for city in cities:
    if city == "laramie":
        print(city.upper())
    else:
        print(city.title())
```

* At the heart of every `if` statement is an expression that can be evaluated

# Checking for Equality
---
* The simplest condition checks whether a variable equals a value
* Use the **double equals** `==` operator — not single `=`!
* `=` is **assignment**: "set the value of city to 'Laramie'"
* `==` **asks a question**: "is the value of city equal to 'Laramie'?"

* Most conditional tests compare the current value of a variable to a value of interest
* The simplest conditions check whether the value of a variable is equal to the value of interest
* Comparisons are done with the double equals `==` operator
* You need to use `==` when doing equality comparisons, as `=` is assignment
* The first line assigns `'Laramie'` to the variable `city` — this is done using `=`
* The following line checks whether the value of city is equivalent to `'Laramie'` using `==`
* The *equality operator* returns `True` if the values on the left and right side match, `False` if they don't
* A single `=` is really a *statement* — you might read it as: "Set the value of city equal to 'Laramie'"
* On the other hand a double equals asks a question: "Is the value of city equal to 'Laramie'?"

```python
city = "Laramie"

# == asks a question and returns True or False
print(city == "Laramie")   # True
print(city == "Cheyenne")  # False
```

# Equality and Case Sensitivity
---
* Testing for equality is **case sensitive** in Python
* If case doesn't matter, convert both sides to the same case before comparing
* `.lower()` doesn't change the stored value — just the value used in the comparison

* Testing for equality is case sensitive in Python
* Two values with different capitalization are not equal
* If case matters this behavior is advantageous
* If case doesn't matter it can be inconvenient
* So what do we do if case doesn't matter and we want to compare elements?
* We are setting the value in the variable and the value we are testing against to be the same case
* `.upper()` doesn't change the value, just the casing
* Websites enforce certain rules for the data that users enter in a similar way
* It could be to ensure your email is correct regardless of the capitalization you entered
* Or to ensure all usernames are truly unique

```python
city = "Laramie"

# Case sensitive — these are NOT equal
print(city == "laramie")  # False

# Fix: convert to same case before comparing
print(city.lower() == "laramie")  # True — works no matter how city is formatted
```

# Checking for Inequality
---
* Use `!=` to check if two values are **not equal**
* If the two variables match, `!=` returns `False`
* If they don't match, `!=` returns `True`
* Many comparisons check for equality, but inequality can also be very useful

* If the two variables match Python returns `False` when using `!=`
* If the two variables do not match `!=` returns `True`
* Many comparisons will check for equality, but inequality can also be useful

```python
topping = "mushrooms"

if topping != "anchovies":
    print("That topping is acceptable!")
```

# Numerical Comparisons
---
Testing numerical values is straightforward   
All the mathematical comparison operators are available:
* `==` equal
* `!=` not equal
* `<` less than
* `>` greater than
* `<=` less than or equal to
* `>=` greater than or equal to

* Testing numerical values is straightforward
* Numerical comparisons can use both `==` and `!=`
* As well as other operators
* Many mathematical comparisons can also be included in conditional statements
	* less than `<`
	* greater than `>`
	* less than or equal to `<=`
	* greater than or equal to `>=`
* Each mathematical comparison can be used within an if statement

```python
age = 19

print(age == 18)  # False
print(age != 18)  # True
print(age > 18)   # True
print(age >= 18)  # True
print(age < 18)   # False
print(age <= 18)  # False
```

# Checking Multiple Conditions
---
* Sometimes you need **two conditions** to both be `True` → use `and`
* Sometimes you only need **one condition** to be `True` → use `or`
* Parentheses aren't required but improve readability

* You may want to check multiple conditions at the same time
* For example you may need two conditions to be `True` to take an action
* Other times you may be satisfied with just one condition being `True`
* The keywords `and` and `or` help in these cases
* Sometimes you may need to check if two statements are both `True` simultaneously
* To do so the `and` keyword is used to combine the two conditionals
* If each passes the overall expression evaluates to `True`
* If either test fails, or both, the expression evaluates to `False`
* To improve readability you can use parentheses — they aren't required but make it more clear
* The `or` keyword allows you to check multiple conditions like `and`
* But this condition passes when one or both of the conditions are true
* An `or` expression only fails when both statements are false

```python
age_0 = 22
age_1 = 18

# and — both must be True
print((age_0 >= 21) and (age_1 >= 21))  # False — age_1 fails
print((age_0 >= 21) and (age_0 >= 18))  # True — both pass
```

```python
age_0 = 22
age_1 = 18

# or — only one needs to be True
print((age_0 >= 21) or (age_1 >= 21))  # True — age_0 passes
print((age_0 < 18) or (age_1 < 18))   # False — neither passes
```

# Checking Membership with `in` and `not in`
---
* Sometimes it's important to check whether a list **contains** a value → use `in`
* Other times check if a value is **not** in a list → use `not in`
* Also works on strings to check if a *substring* exists within

* Sometimes it is important to check whether a list contains a value
* You may want to check to see if a username already exists in a list
* To determine if a value is in a list `in` can be used
* Thinking back to our list of cities, what if we want to add new cities only if they aren't already there
* The keyword `in` tells Python to check for the existence of a value in the list
* This technique is very powerful — it allows you to avoid adding duplicates to your list
* This can also be used on strings to see if a *substring* exists within
* Other times you may want to check to see if an element is *not* in a list
* Suppose you have a list of dis-allowed words
* You can ensure a word is not in the list before allowing it to be posted
* The if statement should be pretty clear: `if "Go Pokes" not in banned_phrases`
* Python returns true if that statement is not in the list

```python
cities = ["Laramie", "Casper", "Cheyenne", "Sheridan"]

print("Laramie" in cities)   # True
print("Gillette" in cities)  # False

# Works on strings too — check for a substring
print("ara" in "Laramie")    # True
```

```python
banned_phrases = ["spam", "click here", "free money"]
message = "Go Pokes!"

if message not in banned_phrases:
    print(f"Message approved: {message}")
```

# Boolean Values
---
* A *boolean expression* is another name for a conditional test
* A **boolean value** is either `True` or `False` — nothing else
* Boolean values are often used to keep track of certain conditions
* They provide an efficient way to track the **state** of a program

* As you learn more about programming you will hear the term *boolean expression*
* This is another name for a conditional test
* A *boolean value* is either `True` or `False`, just like the value of a conditional expression after evaluation
* Boolean values are often used to keep track of certain conditions
* Boolean values provide an efficient way to track the state of a program
* When you understand conditional tests, you can start writing `if` statements
* Several kinds of `if` statements exist
* Your choice of which to use depends on the number of conditions needed to test

```python
# Boolean values track state
game_active = True
can_edit = False

print(game_active)
print(can_edit)

# Conditional tests produce boolean values
age = 20
print(age >= 18)  # True
print(type(age >= 18))  # shows it's a bool
```
