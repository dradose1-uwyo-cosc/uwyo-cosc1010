# Variables
---
Variables in programming serve a greater purpose than what you might think of a variable in math   
We also are then able to do some very *non-mathey* things with them 

Unlike in math, `x = x + 1` is a perfectly valid statement in programming. In math, that equation has no solution. In programming, it means: take the current value of x, add 1 to it, and store that new value back in x. The `=` sign here means assignment, not equality. This trips up a lot of people coming from a math background — totally normal!

```python
x = 5
x = x + 1
```

# Variables

We added a variable to our code named `message`. Every variable is connected to a value — the information associated with that variable. In this case `message` contains the string "Hello World". Adding a variable makes a little more work for the interpreter — when it processes the first line it associates the variable `message` with "Hello World". When it reaches the second line it prints the value associated with `message`. Now let's expand the code to print a second message. You can change the value of a variable in your program at any time — Python will keep track of the current value after each change.

```python
#assign the string "Hello World" to the variable message 
message = "Hello world" 
print(message) 

#overwrite the "Hello world" with 7220
message = 7220
print(message)
```

# Naming and using variables
---
There are some rules for variable names
* They can only contain:
    * Letters
    * Numbers
    * Underscores 
* No spaces
* Avoid keywords (like `print`) 
* Short and descriptive 

When using a variable in Python there are a few rules and guidelines. Breaking some of these rules will cause errors — others are in place just to make code easier to read. Variable names can contain only letters, numbers, and underscores — they must start with a letter or underscore. `message_1` and `_message` are valid, `1_message` is not. Spaces are not allowed — use underscores instead. Avoid using Python keywords and function names like `print` as variable names. Variable names should be short but descriptive. Be careful when using the lowercase letter l and uppercase O as they may look like 1 and 0.
