# What is a String?
---
A string is a **series of characters**    
Anything inside quotes is considered a string    
You can use single *or* double quotes — Python doesn't care which, as long as you're consistent!

Programmers often gather some kind of data and then do something useful with it, so it helps to classify those different kinds of data. The first data type we'll look at is a string. Strings are simple at first glance but can be exceptionally helpful. A string is a series of characters — anything inside quotes is a string in Python. You can use single or double quotes, which allows you to have quotations and apostrophes within strings.

```python
# Both of these are strings
string_one = "This is a string"
string_two = 'This is also a string'

print(string_one)
print(string_two)
```

# Why two kinds of quotes?
---
Having both lets you use quotes and apostrophes *inside* your strings without breaking things    
This is more useful than it might seem at first!

Having both single and double quotes lets you use quotes and apostrophes inside your strings without breaking things. This flexibility is more useful than it might seem at first — especially once we get to strings that contain contractions or quoted speech.

```python
# Single quotes inside double quotes - no problem!
print("The language 'Python' is named after Monty Python, not the snake.")

# Double quotes inside single quotes - also fine!
print('I told my friend, "Python is great!"')
```

# Changing Case
---
Strings have built in **methods** — actions Python can perform on them   
Some of the most useful ones change the case of the text    
A method is called using a **dot** after the variable name

One of the simplest tasks you can do with strings is changing the case of words. The method `title()` appears after the variable in the print call. A method is an action that Python can perform on a piece of data. The dot after `name` in `name.title()` tells Python to make the `title()` method act on the variable `name`. Every method is followed by a set of parentheses — `title()` doesn't need any additional information so its parentheses are empty. There are other useful methods for dealing with case — `upper()` and `lower()`.

```python
name = "ada lovelace"

# title case - capitalizes the first letter of each word
print(name.title())

# all uppercase
print(name.upper())

# all lowercase
print(name.lower())
```

# Why does case matter?
---
To a human, `Ada`, `ADA`, and `ada` are obviously the same name   
To Python, they are **three different strings**    

`.lower()` is particularly useful when storing or comparing data — convert everything to lowercase first so you don't have to worry about what the user typed

The `title()` function changes each word to title case, where each word begins with a capital letter. This is useful because you'll often want your program to recognize the input values Ada, ADA, and ada as the same name and display them all as Ada. The `lower()` method is particularly useful for storing data — often you won't want to trust the casing input by a user, so you'll convert to lowercase before storing. Then when you retrieve the data you can apply the correct casing.

```python
# These look the same to us, but Python sees them as different!
print("Ada" == "ada")
print("Ada" == "ADA")

# Lowercasing first solves this
print("Ada".lower() == "ada".lower())
```

# f-Strings
---
Often you want to combine a variable's value with some text   
**f-strings** make this easy — put an `f` before the opening quote, then wrap variable names in `{}`   
Python replaces the variable name with its actual value

Sometimes you'll want to use a variable's value in a string. For example you may want to use two variables to represent a first and last name, then display those together. To insert a variable's value into a string, place the letter f before the opening quotation mark. Put braces around the name or names of variables you want to use in the string. Python will replace each variable with its value when the string is displayed. These strings are called f-strings — the f is for format.

```python
first_name = "ada"
last_name = "lovelace"

# Combine variables into a full name
full_name = f"{first_name} {last_name}"
print(full_name)

# You can even call methods inside the {}
print(f"Hello, {full_name.title()}!")
```

# f-Strings in practice
---
f-strings are everywhere in real Python code   
They're cleaner than trying to concatenate strings with `+`    
And they make your code much easier to read

Python formats the string by replacing the name of any variable in braces with its value. A lot can be done with f-strings — they can compose complete messages using information associated with a variable, and you can utilize string based methods on variables within f-strings.

```python
university = "University of Wyoming"
mascot = "Cowboys"
year_founded = 1886

# f-strings can include any variable type, not just strings!
print(f"The {university} {mascot} were founded in {year_founded}.")
```

# Whitespace in Strings
---
**Whitespace** refers to any non-printing characters — spaces, tabs, newlines   
Python gives you special characters to add them:
* `\t` — tab
* `\n` — newline (moves to the next line)

Whitespace can be used to organize output to make it easier to read. Tab characters can be added by inserting `\t` into a string. Newline characters can be added with `\n`. As many or as few tabs and newlines can be added to ensure the string is formatted as you want it. The ability to add tabs and newlines are extremely helpful.

```python
# Adding a tab
print("Languages:")
print("\tPython")
print("\tJavaScript")

print() # blank line

# Adding newlines
print("Languages:\nPython\nJavaScript\nC")
```

# Stripping Whitespace
---
Are these the same string?
* `"Python"`
* `"Python "`

To us, yes. To Python, **absolutely not**    
Extra whitespace causes all sorts of problems — especially when comparing user input    
Fortunately Python has methods to clean it up

Extra whitespace can be confusing in programs. To programmers 'python' and 'python   ' look effectively the same. To a program they are two different strings thanks to the trailing whitespace. Python considers the whitespace to be significant unless told otherwise. It is important to think about whitespace — often two strings will need to be compared to see if they're the same. An example would be checking usernames against a stored one to determine equivalency. Python makes it easy to eliminate extra whitespace from data that people enter — Python can look for extra whitespace on the right and left sides of strings.

```python
# rstrip() removes whitespace from the right
string_one = "py    "
string_two = "thon"

print(f"{string_one}{string_two}")          # extra space visible
print(f"{string_one.rstrip()}{string_two}") # stripped!
```

`string_one` contained extra trailing whitespace. When `.rstrip()` is applied that whitespace is removed. It is only removed temporarily — if you asked for the variable `string_one` again it would be back. To remove whitespace from a string permanently you have to associate the stripped version with the variable name. Whitespace can also be stripped from the left with `.lstrip()` or from both sides with `.strip()`.

```python
# lstrip() removes from the left, strip() removes from both sides
string_three = "   python   "

print(f"*{string_three}*")           # shows the spaces
print(f"*{string_three.lstrip()}*")  # left stripped
print(f"*{string_three.rstrip()}*")  # right stripped
print(f"*{string_three.strip()}*")   # both sides stripped
```

# Removing Prefixes
---
Sometimes you want to remove a known chunk from the *start* of a string   
`removeprefix()` does exactly that — and there's a matching `removesuffix()` for the end    
You see this used constantly with URLs!

When working with strings another common task is to remove a prefix. Consider a URL with the common prefix `https://` — suppose you want to remove this prefix to just focus on the URL. This can be done with the `.removeprefix()` method. You enter the name of the variable followed by a dot and the method name. In the parentheses you supply the string you want to remove. Like the methods for removing whitespace, `.removeprefix()` leaves the original unchanged — to make it permanent you can reassign the variable.

```python
url = "https://www.uwyo.edu"

# Strip the https:// prefix
print(url.removeprefix("https://"))

# removesuffix works the same way from the end
filename = "lecture_notes.txt"
print(filename.removesuffix(".txt"))
```

# Syntax Errors with Strings
---
A **syntax error** happens when Python doesn't recognize your code as valid Python    
One common cause: mixing up your quotes    
Python gets confused about where the string starts and ends

One kind of error you will often see is a syntax error. A syntax error occurs when Python doesn't recognize a section of your program as valid. For example if you use an apostrophe inside a single quote it will produce an error — Python assumes the first quotation to the apostrophe is the string, so everything following is an error. To fix this you can use double quotes instead. Using double quotes to encapsulate the string allows the use of single quotes and apostrophes in the string.

```python
# This looks fine but will cause a syntax error!
# The apostrophe in "Python's" ends the string early
message = 'One of Python's strengths is its community.'
print(message)
```

```python
# Fix: use double quotes on the outside when you have an apostrophe inside
message = "One of Python's strengths is its community."
print(message)
```

# Syntax Errors — Don't Panic!
---
Syntax errors are the **most common** errors you'll see as you start out   
They can be frustrating because the error message isn't always obvious   

A tip: if you see a syntax error and can't spot it immediately, look at the line **before** where Python is pointing — the real problem is often one line up

Syntax errors are also the least specific kind of error, so they can be difficult and frustrating to identify and correct. If you get stuck on a particularly stubborn error try taking a break. Your editor's syntax highlighting feature should help you spot some syntax errors quickly as you write your programs — if you see Python code highlighted as if it's English or English highlighted as if it's Python code, you probably have a mismatched quotation mark somewhere.
