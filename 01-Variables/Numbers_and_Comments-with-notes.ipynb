# Numbers in Python
---
Python treats numbers in different ways depending on what they are   
The two main types you'll work with are:
* **Integers** (`int`) — whole numbers, no decimal point
* **Floats** (`float`) — numbers with a decimal point

Knowing which one you have matters more than you might think!

Python treats numbers in several different ways. The two main types are integers — whole numbers with no decimal point — and floats, which are numbers with a decimal point. Knowing which one you have matters more than you might think, as we'll see shortly.

# Integers
---
Integers are the simplest numbers — whole numbers, no decimal point   
You can perform all the basic mathematical operations on them   
Python also supports **exponents** with `**`

Integers are the simplest numbers. In a terminal session Python simply returns the result of an operation. You can add, subtract, multiply, and divide integers in Python. Python uses two multiplication symbols to represent exponents. Python supports order of operations too — multiple operations can be used in a single expression, and you can use parentheses to modify the order of operations.

```python
# Basic operations
print(2 + 3)   # addition
print(10 - 4)  # subtraction
print(3 * 6)   # multiplication
print(10 / 3)  # division
print(3 ** 2)  # exponent (3 squared)
```

```python
# Order of operations works just like math class
print(2 + 3 * 4)    # multiplication happens first
print((2 + 3) * 4)  # parentheses change the order
```

# Floats
---
Any number with a decimal point is a **float**   
For the most part, floats behave exactly as you'd expect   
But there is one quirk worth knowing about...

Python calls any number with a decimal point a float. The term is used in most programming languages — it refers to the fact that a decimal point may appear at any place in the number. Every programming language must be properly designed to manage decimal numbers so numbers behave accordingly. Mostly you can use floats without worry. However, there are some circumstances where you may get an arbitrary number of decimal places in an answer.

```python
# Most float math is perfectly normal
x = 0.1 + 0.1
print(x)  # 0.2 as expected

y = 0.2 + 0.2
print(y)  # 0.4 as expected

z = 2 * 0.1
print(z)  # 0.2 as expected
```

```python
# But sometimes you get something... unexpected
print(0.1 + 0.2)   # this probably isn't what you expected!
print(3 * 0.1)     # same issue
```

# Why does that happen?
---
This isn't a Python bug — it happens in **every** programming language   

Computers store numbers in binary (1s and 0s), and some decimal fractions can't be represented perfectly in binary — just like how `1/3` can't be written perfectly as a decimal (0.333333...)    

For now: just be aware it exists, and don't panic when you see it!

This happens in all languages and is of little concern. Python tries to find a way to represent the result as precisely as possible — this is difficult given how computers represent numbers internally. Just ignore the extra decimal places for now — you'll learn ways to deal with them when needed. The important thing is knowing it exists so you don't panic when you see it.

# Integers and Floats Together
---
You can mix integers and floats freely in Python   
But there's an important rule: **any operation involving a float gives you a float**   
Even division of two integers gives a float!

When you divide any two numbers you will get a float — even if dividing two integers and the result is a whole number. If you mix an integer and float in any operation you will get a float. Python defaults to a float in any operation involving a float, regardless of whether the output is a whole number.

```python
# Division always returns a float
print(4 / 2)    # you might expect 2, but you get 2.0

# Mixing int and float gives float
print(1 + 2.0)  # 3.0
print(2 * 3.0)  # 6.0
print(3.0 ** 2) # 9.0
```

# Underscores in Numbers
---
Large numbers can be hard to read — how many zeros does `14000000000` have?   
Python lets you use underscores as visual separators, like commas in regular writing   
Python completely ignores them — they're just for *you*

When writing long numbers you can group digits using underscores to make large numbers more readable — `universe_age = 14_000_000_000`. When a number that was defined using underscores is printed, only the digits are shown. Python ignores the underscores when storing these kinds of values. This applies even if a grouping of three isn't used — 1000 is the same as 1_000 which is the same as 10_00.

```python
# Hard to read
universe_age = 14000000000
print(universe_age)

# Much easier to read — Python ignores the underscores
universe_age = 14_000_000_000
print(universe_age)
```

# Multiple Assignment
---
Python lets you assign values to multiple variables in a single line   
This is handy when initializing several related variables at once   
Just make sure the number of variables matches the number of values!

You can assign values to more than one variable using just a single line. This can help shorten your programs while also making them easier to read. You just need to separate the values and variable names by a comma. Python will assign each value to its respective variable as long as the number of items on each side matches.

```python
# Assign x, y, and z all at once
x, y, z = 0, 0, 0
print(x, y, z)

# Can also assign different values
length, width, height = 10, 5, 3
print(f"Volume: {length * width * height}")
```

# Constants
---
A **constant** is a variable whose value should never change   
Python doesn't actually enforce this — nothing stops you from changing it   
Instead, Python programmers use a **convention**: write constants in ALL_CAPS   
This is a signal to yourself and others: *don't change this*

A constant is a variable whose value stays the same throughout program execution. Python doesn't have a built in constant type — Python programmers use all capital letters to indicate a variable should be treated as a constant and never changed. When you want to treat a variable as a constant in your code, write the name in all capital letters.

```python
# Constants by convention — all caps tells everyone "don't change this"
MAX_CONNECTIONS = 5000
PI = 3.14159
SCHOOL_NAME = "University of Wyoming"

print(f"Max connections allowed: {MAX_CONNECTIONS}")
print(f"Approximate value of pi: {PI}")
print(f"School: {SCHOOL_NAME}")
```

# Comments
---
A **comment** is a note you write in your code for humans to read — Python ignores it completely   
Start a comment with `#` — everything after it on that line is ignored   

You've already seen comments in the code cells above!

Comments are an extremely useful feature in most programming languages. Everything you've written so far is Python code — as your programs become longer and more complicated you should add notes within them that describe your overall approach. A comment allows you to write notes in your spoken language within your programs. In Python the hash mark # indicates a comment. Anything following a hash mark in your code is ignored by the Python interpreter.

```python
# This is a comment — Python ignores this line entirely
print("Hello!")  # You can also add a comment at the end of a line

# Comments can explain WHAT your code does
# Calculate the area of a rectangle
length = 10
width = 5
area = length * width
print(f"Area: {area}")
```

# Why bother with comments?
---
Right now your programs are short enough that you can remember what everything does   
But come back to your code in two weeks and see how well that holds up!   

Good comments explain **why** you made a decision, not just *what* the code does   
Think of them as notes to your future self

The main reason to write comments is to explain what your code is supposed to do and how you are making it work. When you're in the middle of working on a project you understand how all the pieces fit together. But when you return to a project after some time away you'll likely have forgotten some details. Writing good comments can save you time by summarizing your overall approach. If you want to become a professional programmer or collaborate with others, you should write meaningful comments. Writing clear, concise comments is one of the most beneficial habits you can form as a new programmer.

```python
x = 5

# Bad comment — just describes what the code already shows
x = x + 1  # add 1 to x

# Better comment — explains WHY
x = x + 1  # increment counter to move to the next student record
```
