# Numbers in Python
---
Python treats numbers in different ways depending on what they are   
The two main types you'll work with are:
* **Integers** (`int`) — whole numbers, no decimal point
* **Floats** (`float`) — numbers with a decimal point

Knowing which one you have matters more than you might think!

# Integers
---
Integers are the simplest numbers — whole numbers, no decimal point   
You can perform all the basic mathematical operations on them   
Python also supports **exponents** with `**`

```python
# Basic operations
print(2 + 3)   # addition
print(10 - 4)  # subtraction
print(3 * 6)   # multiplication
print(10 / 3)  # division
print(3 ** 2)  # exponent (3 squared)
```

```python
# Order of operations works just like math class
print(2 + 3 * 4)    # multiplication happens first
print((2 + 3) * 4)  # parentheses change the order
```

# Floats
---
Any number with a decimal point is a **float**   
For the most part, floats behave exactly as you'd expect   
But there is one quirk worth knowing about...

```python
# Most float math is perfectly normal
x = 0.1 + 0.1
print(x)  # 0.2 as expected

y = 0.2 + 0.2
print(y)  # 0.4 as expected

z = 2 * 0.1
print(z)  # 0.2 as expected
```

```python
# But sometimes you get something... unexpected
print(0.1 + 0.2)   # this probably isn't what you expected!
print(3 * 0.1)     # same issue
```

# Why does that happen?
---
This isn't a Python bug — it happens in **every** programming language   

Computers store numbers in binary (1s and 0s), and some decimal fractions can't be represented perfectly in binary — just like how `1/3` can't be written perfectly as a decimal (0.333333...)    

For now: just be aware it exists, and don't panic when you see it!

# Integers and Floats Together
---
You can mix integers and floats freely in Python   
But there's an important rule: **any operation involving a float gives you a float**   
Even division of two integers gives a float!

```python
# Division always returns a float
print(4 / 2)    # you might expect 2, but you get 2.0

# Mixing int and float gives float
print(1 + 2.0)  # 3.0
print(2 * 3.0)  # 6.0
print(3.0 ** 2) # 9.0
```

# Underscores in Numbers
---
Large numbers can be hard to read — how many zeros does `14000000000` have?   
Python lets you use underscores as visual separators, like commas in regular writing   
Python completely ignores them — they're just for *you*

```python
# Hard to read
universe_age = 14000000000
print(universe_age)

# Much easier to read — Python ignores the underscores
universe_age = 14_000_000_000
print(universe_age)
```

# Multiple Assignment
---
Python lets you assign values to multiple variables in a single line   
This is handy when initializing several related variables at once   
Just make sure the number of variables matches the number of values!

```python
# Assign x, y, and z all at once
x, y, z = 0, 0, 0
print(x, y, z)

# Can also assign different values
length, width, height = 10, 5, 3
print(f"Volume: {length * width * height}")
```

# Constants
---
A **constant** is a variable whose value should never change   
Python doesn't actually enforce this — nothing stops you from changing it   
Instead, Python programmers use a **convention**: write constants in ALL_CAPS   
This is a signal to yourself and others: *don't change this*

```python
# Constants by convention — all caps tells everyone "don't change this"
MAX_CONNECTIONS = 5000
PI = 3.14159
SCHOOL_NAME = "University of Wyoming"

print(f"Max connections allowed: {MAX_CONNECTIONS}")
print(f"Approximate value of pi: {PI}")
print(f"School: {SCHOOL_NAME}")
```

# Comments
---
A **comment** is a note you write in your code for humans to read — Python ignores it completely   
Start a comment with `#` — everything after it on that line is ignored   

You've already seen comments in the code cells above!

```python
# This is a comment — Python ignores this line entirely
print("Hello!")  # You can also add a comment at the end of a line

# Comments can explain WHAT your code does
# Calculate the area of a rectangle
length = 10
width = 5
area = length * width
print(f"Area: {area}")
```

# Why bother with comments?
---
Right now your programs are short enough that you can remember what everything does   
But come back to your code in two weeks and see how well that holds up!   

Good comments explain **why** you made a decision, not just *what* the code does   
Think of them as notes to your future self

```python
# Bad comment — just describes what the code already shows
x = x + 1  # add 1 to x

# Better comment — explains WHY
x = x + 1  # increment counter to move to the next student record
```

# Numbers and Comments — Summary
---
* **Integers** are whole numbers, **floats** have decimal points
* Python supports `+`, `-`, `*`, `/`, `**` and follows order of operations
* Division always returns a float — even `4 / 2` gives `2.0`
* Floats occasionally have tiny precision quirks — this is normal and happens everywhere
* Use underscores in large numbers to make them readable: `14_000_000_000`
* **Constants** are written in ALL_CAPS by convention
* **Comments** start with `#` and are ignored by Python — write them for humans!
