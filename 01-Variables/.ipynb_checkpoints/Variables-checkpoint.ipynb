# Variables
---
Variables can hold any type we might need or want    
Sometimes they hold numbers, represented by `int` or `float`   
Othertimes they may hold collections of characters known as `strings`

```python
#assign the string "Hello World" to the variable message 
message = "Hello world" 
print(message) 

#overwrite the "Hello world" with 7220
message = 7220
print(message)
```

# Naming and using variables
---
There are some rules for variable names
* They can only contain:
    * Letters
    * Numbers
    * Underscores 
* No spaces
* Avoid keywords (like `print`) 
* Short and descriptive
