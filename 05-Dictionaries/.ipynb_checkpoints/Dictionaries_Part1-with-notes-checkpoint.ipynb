# Working with Dictionaries
---
* A collection of **key-value** pairs
* A key can be essentially anything
* Key-value pairs are connected with a colon `key:value`
* Dictionaries use **braces** `{}`

```python
states = {'Wyoming': 'Cheyenne', 'Colorado': 'Denver'}
```

* A `dictionary` in Python is a collection of *key-value* pairs
* Each *key* is connected to a *value*
* A key's value can be a number, a string, a list, or even another dictionary
* In python a dictionary is wrapped in braces `{}` with a series of `key:value` pairs inside
	* `states = {'Wyoming':'Cheyenne','Colorado':'Denver'}`
* A `key:value` pair is a set of values associated with each other
* When you provide a key, Python returns its associated value
* Every `key:value` pair is connected with a colon `:`
* The simplest dictionary contains only one pair
* But, dictionaries can store as many pairs as you want

````python
# A simple dictionary mapping states to their capitals
states = {'Wyoming': 'Cheyenne', 'Colorado': 'Denver'}
print(states)
````

# Accessing Values
---
* Values are accessed with their **key** — not an index position
* Give the name of the dictionary and place the key in **square brackets**
* `states['Wyoming']` returns `'Cheyenne'`

* To get the value associated with a key, give the name of the dictionary and place the key in square brackets
	* `states['Wyoming']`
* That returns the value associated with the key, so `"Cheyenne"`
* You can have an unlimited number of `key:value` pairs in a dictionary
* You can access any of the values with their corresponding key
* Use the square brackets to retrieve the values, rather than braces

````python
states = {'Wyoming': 'Cheyenne', 'Colorado': 'Denver', 'Montana': 'Helena'}

# Access a value by its key
print(states['Wyoming'])
print(states['Colorado'])

# Use in an f-string
print(f"The capital of Wyoming is {states['Wyoming']}.")
````

# Adding New Key-Value Pairs
---
* Dictionaries are **dynamic** structures, like lists
* You can add new `key:value` pairs at any time
* Syntax: `dictionaryName[newKey] = newValue`
* Dictionaries retain the order in which they were defined

* Dictionaries are dynamic structures, like lists
* You can add new `key-value` pairs to a dictionary at any time
* To add a new `key-value` pair you give the name of the dictionary followed by the key in square brackets, assigning the new value
* So a new state could be added with `states["Montana"] = "Helena"`
* There we defined our states dictionary again
* We printed it out to verify the two states were there
* Then supplying a new key we assign a new value to the list
* Dictionaries retain the order in which they were defined

````python
states = {'Wyoming': 'Cheyenne', 'Colorado': 'Denver'}
print(states)

# Add a new key:value pair
states['Montana'] = 'Helena'
states['Utah'] = 'Salt Lake City'
print(states)
````

# Starting with an Empty Dictionary
---
* Sometimes it's helpful to start with an **empty dictionary**
* Define one with empty braces: `states = {}`
* Then add `key:value` pairs as needed
* Typically used to store user-supplied data, or built programmatically

* Sometimes it is helpful to start with an empty dictionary
* Then as needed adding new elements to it
* To start filling an empty dictionary, define a dictionary with an empty set of braces
	* `states = {}`
* Then, each `key:value` pair can be added
* Typically empty dictionaries are used to store user supplied data
* Can be done programmatically

````python
# Start empty and build up
states = {}

states['Wyoming'] = 'Cheyenne'
states['Colorado'] = 'Denver'
states['Montana'] = 'Helena'

print(states)
````

# Modifying Values
---
* Modifying a value follows the **same syntax** as adding one
* Use the dictionary name and key in square brackets, then assign a new value
* Be careful not to accidentally overwrite an existing value when you mean to add a new one!

* Modifying a value in a dictionary follows the same syntax as adding a value
* The dictionary name and key in square brackets with an assignment to it
* You need to be careful to not overwrite already existing values when trying to declare a new one

````python
# A dictionary about UW
uw = {
    'name': 'University of Wyoming',
    'mascot': 'Pistol Pete',
    'founded': 1886,
    'location': 'Laramie'
}
print(uw)

# Modify an existing value
uw['mascot'] = 'Cowboys'
print(uw)
````

# Removing Key-Value Pairs
---
* When you no longer need information, use the `del` statement
* `del` needs the dictionary name and the key to remove
* The deleted `key:value` pair is removed **immediately** and permanently

* When you no longer need information in a dictionary, the `del` statement can be used
* All `del` needs is the name of the dictionary and the key you want to remove
* So to remove a state: `del states['Colorado']`
* The `del` statement tells python to delete the key from the dictionary
* It also removes the value associated with that key
* The deleted `key:value` pair is removed immediately

````python
states = {'Wyoming': 'Cheyenne', 'Colorado': 'Denver', 'Montana': 'Helena'}
print(states)

# Remove a key:value pair
del states['Colorado']
print(states)
````

# A Dictionary of Similar Objects
---
* Previous examples stored similar kinds of information (state: capital)
* You can also use dictionaries to store **multiple pieces of info about one object**
* Dictionary declarations don't need to be on one line — new lines between entries is fine!

* Our previous examples were used to store similar kinds of information
	* e.g. States and their Capitals
* You could also use dictionaries to store multiple pieces of information about one object
* Dictionary declarations do not need to be declared all out on one line
* You can have new lines between entries

````python
# Multiple pieces of info about one object
uw = {
    'name': 'University of Wyoming',
    'mascot': 'Cowboys',
    'founded': 1886,
    'location': 'Laramie',
    'enrollment': 12000,
}

print(f"{uw['name']} was founded in {uw['founded']} in {uw['location']}.")
print(f"Their teams are known as the {uw['mascot']}.")
````

# Using `get()` to Access Values Safely
---
* Using square brackets has one potential problem: if the key doesn't exist you get a `KeyError`
* `get()` provides a safer alternative
* First argument: the key
* Second argument (optional): what to return if the key isn't found

* Using keys in square brackets to retrieve values has one potential problem:
	* If the key doesn't exist you will get an error
* Namely a `KeyError`
* Alternatively `get()` can be used
* The `get()` method requires a key as its first argument
* It also takes a second argument optionally, what will be returned if no corresponding key is found

````python
states = {'Wyoming': 'Cheyenne', 'Colorado': 'Denver'}

# This will cause a KeyError — the key doesn't exist!
print(states['Idaho'])
````

````python
states = {'Wyoming': 'Cheyenne', 'Colorado': 'Denver'}

# get() handles missing keys gracefully
print(states.get('Wyoming', 'State not found'))   # key exists
print(states.get('Idaho', 'State not found'))      # key missing — returns default
````

# Dictionaries Part 1 — Summary
---
* A **dictionary** is a collection of `key:value` pairs, wrapped in `{}`
* Access values with `dictionary['key']` — not an index
* Add pairs with `dictionary['newKey'] = newValue`
* Modify pairs the same way — overwrite by key
* Remove pairs with `del dictionary['key']`
* Use `get()` to safely access values and avoid `KeyError`
