# Two Ways to Plot
---
Matplotlib has two interfaces — both are common in the wild:

1. **pyplot (global state)** — quick and simple, `plt.plot()` directly
2. **Object-oriented (figures and axes)** — more control, better for complex plots

We'll start with pyplot for simplicity, then show the OOP approach

* The `pyplot` module provides a global state that can be manipulated
* This is intended to be used for interactive data exploration
* It is best when creating simple plots, usually with single figures
* `pyplot` is convenient and many examples out in the world use it
* For more complex plotting Matplotlib's object-oriented API comes into its own

# Basic Line Plot
---
* `import matplotlib.pyplot as plt` — standard convention
* `plt.plot(x, y)` — create a line plot
* Always add `plt.show()` to display the plot
* Add titles and labels to make the chart readable

* The `pyplot` module provides a global state that can be manipulated
* This is intended to be used for interactive data exploration
* Before being able to demonstrating plotting data, data is needed
* Matplotlib's defaults are generally less than ideal
* One of its greatest strengths is the sheer amount of customization that can be performed
* Adding a title and label for the axes is easy
* Typically your figures need to say what the lines mean, among other things

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Basic line plot
plt.plot(x, y)
plt.title('Sine Wave')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
plt.show()
```

# Multiple Lines and Legends
---
* Call `plt.plot()` multiple times to add more lines
* Use the `label=` argument to name each line
* Call `plt.legend()` to display the legend
* Use `loc='best'` to let Matplotlib find the best position

* Typically your figures need to say what the lines mean, among other things
* Matplotlib has a handy legend box for line labeling
	* It like most things are heavily configurable
* The `legend` method can:
	* Set labels
	* Suggest a location for the legend box
	* Configure a number of other things
* Sets the labels for the plots
* Uses the best location for the legend to avoid line overlap

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.plot(x, np.sin(x) * np.cos(x), label='sin(x)*cos(x)')

plt.title('Trig Functions')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='best')
plt.grid(True)
plt.show()
```

# Object-Oriented Approach: Figures and Axes
---
* A **Figure** is the drawing area — can contain multiple plots
* **Axes** are the individual plots within a figure
* `fig, ax = plt.subplots()` creates both at once
* Use `ax.plot()`, `ax.set_title()`, `ax.set_xlabel()` etc.

* If you want to have more control over the charts, Matplotlib's `figure` and `axes` OOP approach is better
* Most advanced plotting demos utilize this
* In essence with OOP Matplotlib you deal with a `figure`
	* A figure is essentially the drawing area with one or more axes (or plots) embedded within
* Both `figure` and `axes` have properties that can be independently specified
* Calling `subplots` without arguments returns a figure with single axes
* This is the closest to using the `Pyplot` state machine
* Calling `subplots` with arguments for numbers of rows (`nrows`) and columns (`ncolumns`) allows multiple plots to be placed in a grid layout
* The call to `subplots` returns the `figure` and an array of `axes` in row-column order

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

# OOP approach — create figure and axes explicitly
fig, ax = plt.subplots(figsize=(8, 4), dpi=100)

ax.plot(x, np.sin(x), label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')

ax.set_title('Trig Functions (OOP style)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend(loc='best')
ax.grid(True)

plt.show()
```

# Scatter Plots
---
* `plt.scatter(x, y)` or `ax.scatter(x, y)`
* Useful for showing relationships between two variables
* Can customize **size** and **color** of individual points
* Matplotlib autoscales the axes automatically

* Scatter plots are another useful chart
* They take a 2D arrays of points, with options for size, color, and more
* Matplotlib is able to autoscale the x and y limits
* The size and color of individual points can be adjusted by passing an array of marker sizes and color indices to the current color map
* We are specifying the area of the marker's bounding boxes, not the circles' diameters
* So if you want the diameter to double, then you need increase the points by a factor of four

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

# Generate fake student data
study_hours = np.random.uniform(1, 10, 50)
grades = study_hours * 8 + np.random.normal(0, 5, 50)
grades = np.clip(grades, 40, 100)  # keep grades in range

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(study_hours, grades, alpha=0.6, color='steelblue')
ax.set_title('Study Hours vs Grade')
ax.set_xlabel('Hours Studied per Week')
ax.set_ylabel('Grade')
ax.grid(True, alpha=0.3)

plt.show()
```

# Bar Charts
---
* A staple for comparing categories
* `ax.bar(labels, values)` for vertical bars
* `ax.barh(labels, values)` for horizontal bars
* Customize colors, width, and labels

* The humble bar chart is a staple for a lot of visual data exploration
* As with most Matplotlib charts, there is space for customization
* There are also a few variants
* Often you will have to supply your own bar and label locations
* More than just line plots can be created
* Matplotlib has a number of plots available

```python
import matplotlib.pyplot as plt
import numpy as np

# Wyoming city populations
cities = ['Cheyenne', 'Casper', 'Laramie', 'Sheridan', 'Gillette']
populations = [65000, 58000, 32000, 17000, 32000]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(cities, populations, color='steelblue', edgecolor='white')

# Add value labels on top of each bar
for bar, pop in zip(bars, populations):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500,
            f'{pop:,}', ha='center', va='bottom', fontsize=9)

ax.set_title('Wyoming City Populations')
ax.set_xlabel('City')
ax.set_ylabel('Population')
ax.set_ylim(0, 75000)

plt.show()
```

# Putting It All Together
---
The complete pipeline: **data → Pandas → Matplotlib**

Read data with Pandas, compute what you need, visualize with Matplotlib

This is what makes Python so powerful for data analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create a dataset
np.random.seed(42)
df = pd.DataFrame({
    'major': np.random.choice(['CS', 'Biology', 'Business', 'Psychology'], 100),
    'grade': np.random.randint(60, 100, 100),
    'study_hours': np.random.uniform(1, 15, 100)
})

# Compute average grade by major using Pandas
avg_grades = df.groupby('major')['grade'].mean().sort_values(ascending=False)

# Visualize with Matplotlib
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Bar chart — average grades by major
axes[0].bar(avg_grades.index, avg_grades.values, color='steelblue')
axes[0].set_title('Average Grade by Major')
axes[0].set_xlabel('Major')
axes[0].set_ylabel('Average Grade')
axes[0].set_ylim(0, 100)

# Scatter — study hours vs grade
axes[1].scatter(df['study_hours'], df['grade'], alpha=0.4, color='steelblue')
axes[1].set_title('Study Hours vs Grade')
axes[1].set_xlabel('Study Hours per Week')
axes[1].set_ylabel('Grade')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# Saving Figures
---
* Save a plot to a file with `plt.savefig('filename.png')`
* Supported formats: PNG, PDF, SVG, EPS
* Set resolution with `dpi=` — higher DPI = better quality
* Call `savefig()` **before** `show()` — show clears the figure

* Matplotlib shines at saving plots, providing many output formats
* The available formats depend on the back-ends available, but generally:
	* PNG, PDF, PS, EPS, SVG
* Saving is as simple as `plt.savefig("filename.whatever")`
* You can set the format explicitly with the `format=` keyword argument
* Matplotlib is used to deliver publication-quality images
* By default, a point is approximately 1/72 of an inch wide
* Matplotlib allows this to be adjusted with dots-per-inch (dpi)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.plot(x, np.sin(x), color='steelblue', linewidth=2)
ax.set_title('Sine Wave')
ax.set_xlabel('x')
ax.set_ylabel('sin(x)')
ax.grid(True, alpha=0.3)

# Save BEFORE show() — show() clears the figure!
plt.savefig('sine_wave.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved to sine_wave.png")
```
