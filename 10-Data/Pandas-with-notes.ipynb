# The DataFrame
---
* Pandas' core data structure — think of it as a **programmable spreadsheet**
* Has **rows** (with an index) and **columns** (with labels)
* Can hold different data types in different columns — unlike NumPy arrays
* By convention, DataFrame variable names start with `df`

* The first step in a Pandas session is to load some data
* There are various ways to do so, and data can be loaded from multiple sources
* By convention `DataFrame` objects' name starts with `df`
* The easiest way to create a `DataFrame` is to use a Python dictionary
	* Though it won't often be used
	* Typically the data will be read in from a file
* Pandas has a large amount of `read_[format]` or `to_[format]` methods
* It covers most data-loading use cases, from CSV to databases
* By default Pandas will try to convert the loaded data sensibly

```python
import pandas as pd

# Create a DataFrame from a dictionary
data = {
    'city': ['Cheyenne', 'Casper', 'Laramie', 'Sheridan', 'Gillette'],
    'population': [65000, 58000, 32000, 17000, 32000],
    'county': ['Laramie', 'Natrona', 'Albany', 'Sheridan', 'Campbell'],
    'elevation_ft': [6062, 5150, 7220, 3743, 4544]
}

df = pd.DataFrame(data)
print(df)
```

# Inspecting a DataFrame
---
* `.head()` — shows the first 5 rows (pass a number to change)
* `.info()` — column names, data types, non-null counts
* `.describe()` — summary statistics for numerical columns
* `.columns` — list of column names
* `.shape` — (rows, columns)

* A quick way to get the row-column structure of a `DF` is to use the `head` method
* The head method shows the first five elements
* The `df`'s columns are indexed by the `column` property
	* It is a `Pandas` `index` instance
	* The columns can be accessed and selected
* Initially Pandas `rows` have a single numeric index

```python
import pandas as pd
import numpy as np

# Use a larger dataset — generate some random student data
np.random.seed(42)
df = pd.DataFrame({
    'student_id': range(1, 21),
    'grade': np.random.randint(60, 100, 20),
    'major': np.random.choice(['CS', 'Biology', 'Business', 'Psychology'], 20),
    'year': np.random.choice([1, 2, 3, 4], 20)
})

print("First 5 rows:")
print(df.head())
print(f"\nShape: {df.shape}")
print(f"\nColumns: {list(df.columns)}")
```

```python
# Summary statistics
print(df.describe())
print()
print(df.info())
```

# Accessing Columns and Rows
---
* Access a **column** with dot notation or `df['column_name']` — returns a **Series**
* Access **rows** by label with `df.loc[label]`
* Access **rows** by position with `df.iloc[position]`
* A **Series** is a one-dimensional labeled array — the building block of DataFrames

* You can grab a column using `DataFrame` dot notation, or the conventional array access via keyword
* It will return a Pandas `series`
	* With all the column fields and their `DataFrame` indices preserved
* There are two main ways to select a row from the `DataFrame`
	* The `loc` method which selects by index label
	* The `iloc` method which selects by position
* The rows and columns of a `DataFrame` are stored as `Pandas Series`
	* A heterogeneous counterpart of NumPy arrays
* These are essentially one-dimensional arrays that can contain any datatype

```python
# Access a column — returns a Series
print("Grade column:")
print(df['grade'].head())

print("\nMajor column (dot notation):")
print(df.major.head())

# Access rows by position
print("\nFirst row (iloc):")
print(df.iloc[0])

# Access rows by label
print("\nRow with index 2 (loc):")
print(df.loc[2])
```

# Filtering and Selecting Data
---
* Filter rows using **boolean conditions** — similar to NumPy
* Select multiple columns with a list: `df[['col1', 'col2']]`
* `groupby()` groups data by a column for aggregate operations

* There are various ways to select groups (subset of rows)
* This operation will occur on a `DataFrame` and return a filtered `DataFrame`
* Often it will be a selection of all columns with a specified column value
* One way to accomplish this is utilizing the `groupby` method
	* This groups a column (or list of)
* `get_group` method allows you to select the desired group

```python
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'student_id': range(1, 21),
    'grade': np.random.randint(60, 100, 20),
    'major': np.random.choice(['CS', 'Biology', 'Business', 'Psychology'], 20),
    'year': np.random.choice([1, 2, 3, 4], 20)
})

# Filter rows where grade > 85
high_grades = df[df['grade'] > 85]
print(f"Students with grade > 85:\n{high_grades}")

print()

# Group by major and get mean grade
print("Average grade by major:")
print(df.groupby('major')['grade'].mean().round(1))
```
