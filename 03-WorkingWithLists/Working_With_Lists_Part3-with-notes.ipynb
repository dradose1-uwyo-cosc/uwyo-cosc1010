# Slices
---
* A *slice* is a specific group of items from a list
* Specify the index of the **first** and **last** elements you want
* Like `range()`, Python stops **one item before** the upper bound
* Any subset can be generated — front, middle, or end

* To make a slice, you specify the index of the first and last elements you want to work with
* As with the `range()` function, Python stops one item before the upper bound
* So, to get the first three items in a list you would request indices 0 through 3
* This would return elements 0, 1, 2 — again, upper bound is exclusive
* The output retains the structure of a list
* Any subset can be generated, so you can take elements from the center as well as the ends

```python
states = ["Alabama", "Alaska", "Arizona", "Arkansas", "California",
          "Colorado", "Connecticut", "Delaware", "Florida", "Georgia"]

# First three states (indices 0, 1, 2)
print(states[0:3])

# Middle of the list
print(states[2:6])

# Last three states
print(states[7:10])
```

# Omitting Slice Bounds
---
* Omit the **first** bound → Python starts at the beginning
    * `states[:3]` is equivalent to `states[0:3]`
* Omit the **second** bound → Python goes to the end of the list
    * `states[2:]` gets everything from index 2 onward
* Use **negative indices** to get the final `x` elements: `states[-3:]`

* If you omit the first bound Python assumes you want to start at the front
	* `states[:3]` is equivalent to `states[0:3]`
* A similar syntax works with the end of the list
* If you want to include all elements starting part way through to the end you can omit the second bound
	* `states[47:]`
* This syntax allows you to output all elements from any point in the list
* This applies regardless of list size
* Recall that a negative number allows us to access a certain distance from the end of the list
* You can use this same idea to access the final `x` amount of elements in a list
* So if you want the final three states: `states[-3:]`

```python
states = ["Alabama", "Alaska", "Arizona", "Arkansas", "California",
          "Colorado", "Connecticut", "Delaware", "Florida", "Georgia"]

print(states[:3])   # first three — same as states[0:3]
print(states[7:])   # from index 7 to the end
print(states[-3:])  # last three — works regardless of list size!
```

# Looping Through a Slice
---
* You can use a slice in a `for` loop to loop through only a **subset** of elements
* Works exactly the same as a normal loop — just include the slice bounds
* Useful for processing data in chunks or displaying paginated results

* You can use a slice in a `for` loop
* This lets you loop through only a subset of elements
* This works in the same way as a normal loop would
* The difference is that where we would normally only have the list name, we include the slice bounds
* Slices are useful in a number of situations
* When you're working with data you can use slices to process the data in chunks
* Or if you're building web applications you can use slices to display information

```python
states = ["Alabama", "Alaska", "Arizona", "Arkansas", "California",
          "Colorado", "Connecticut", "Delaware", "Florida", "Georgia"]

print("The first five states:")
for state in states[:5]:
    print(f"  {state}")
```

# Copying a List
---
* Often you may start with an existing list and want to make a new list based on it
* To copy a list use a slice that includes the **entire** original list: `list[:]`
* **Why not just** `new_list = original`?
    * That doesn't make a copy — it makes a **second name for the same list**
    * Changes to one will be reflected in the other!

* Often you may start with an existing list and may make a new list based on the first
* But how does copying a list in Python work?
* To copy a list you can make a slice that includes the entire original list
* This is done by omitting the bounds, only including the colon
	* `new_states = states[:]`
* But why not `new_states = states`?
* `new_states = states` does not make a true copy of the list
* It instead makes a second name for the same list
* Meaning any changes we made to `states` would be reflected in `new_states` and vice versa
* We are just associating a new variable with the same value
* So to make a new list based on the original we want to use the slice method
* First we make our new variable `new_states`
* Then we assign the whole slice of `states` to it: `new_states = states[:]`

```python
# The WRONG way to copy — both variables point to the same list!
states = ["Wyoming", "Colorado", "Montana"]
other_states = states

states.append("Idaho")
print("states:", states)
print("other_states:", other_states)  # Idaho appears here too — not what we wanted!
```

```python
# The RIGHT way — use [:] to make a true copy
states = ["Wyoming", "Colorado", "Montana"]
other_states = states[:]

states.append("Idaho")
print("states:", states)
print("other_states:", other_states)  # Idaho does NOT appear — true independent copy!
```

# Tuples
---
* Lists work well for collections that change
* Sometimes you want a list of items that **doesn't change**
* Python calls items that cannot change **immutable**
* An immutable list is called a **tuple**
* A tuple looks just like a list but uses **parentheses** instead of square brackets

* Lists work well for storing collections of items
* These collections may change throughout the run of a program
* The ability to modify lists is a particularly important feature
* Sometimes you may want to create a list of items that doesn't change
* *Tuples* allow you to do just that
* Python refers to items that cannot change as *immutable*, and an *immutable* list is a *tuple*
* A tuple looks just like a list
* The difference is that it uses parentheses rather than square brackets
* Once a tuple is defined individual items can be accessed using their index
* That is the same as you would for a list
* Suppose you want to set dimensions for a rectangle that never change

```python
# A tuple uses parentheses — these dimensions should never change
dimensions = (1920, 1080)

print(dimensions[0])  # access by index just like a list
print(dimensions[1])
```

# Tuples are Immutable
---
* What happens if you try to change a tuple?
* Python kicks back a `TypeError` — this is exactly what we want!
* Tuples are really defined by the **commas** — parentheses are for clarity
* A single-element tuple needs a trailing comma: `my_tuple = (1886,)`

* We defined a tuple dimensions
* It uses parentheses rather than a square bracket
* Each item can be printed separately or together from the tuple
* But, tuples are supposed to be *immutable* — what happens if we try to change one?
* We can try to assign a new value to the second element
* However, Python will kick back an error when a line tries to change it
* This is exactly what we want
* Tuples are really defined by the commas
* The parentheses are unneeded, but better for clarity
* You can define a tuple with a single element by having a trailing comma
* `my_tuple = 1886,`

```python
dimensions = (1920, 1080)

# Try to change a value — Python won't allow it!
dimensions[0] = 1280
```

# Looping Through a Tuple
---
Looping through a tuple works **exactly** the same as looping through a list!

* Looping through a tuple is easy
* It is done exactly in the same fashion as lists!
* Tuples can't be modified, as we saw
* They can however have a new value assigned to the variable holding them
* For example to change our `dimensions` tuple we can redefine a new tuple
* We have our initial tuple and can print it for viewing
* We then associate a new tuple with the variable `dimensions`
* Python has no qualms as reassignments are fine
* When you compare with lists, tuples are simple data structures
* Use them if you want to store a set of values that should not be changed

```python
dimensions = (1920, 1080)

for dimension in dimensions:
    print(dimension)
```

```python
# You can't change a tuple — but you CAN reassign the variable to a NEW tuple
dimensions = (1920, 1080)
print("Original:", dimensions)

dimensions = (2560, 1440)  # This is fine — we're replacing the tuple, not modifying it
print("New:", dimensions)
```

# Slices, Copying, Tuples — Summary
---
* **Slices** let you work with a subset of a list: `list[start:end]`
* Omitting bounds: `[:3]` starts at beginning, `[3:]` goes to end, `[-3:]` gets last three
* Use slices in `for` loops to iterate over a subset
* **Copying**: use `list[:]` to make a true independent copy — `new = old` does NOT copy!
* **Tuples** use `()` instead of `[]` and are **immutable** — values cannot be changed
* Loop through tuples exactly like lists
* Use tuples when you want to protect values that should never change
