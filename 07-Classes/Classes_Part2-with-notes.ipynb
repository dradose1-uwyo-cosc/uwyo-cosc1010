Inheritance lets you **build on what already exists** rather than starting from scratch

* One of the first tasks being modifying the associated attributes

# The Car Class
---
* A more realistic example than dogs for most students!
* Stores `make`, `model`, `year` as attributes
* Includes a method to display a descriptive name
* We'll build on this class throughout the video

* We can write a new class, this time representing a car
* This class will store information relating to a car
* As well as a method to summarize information
* In the car class we define the `__init__()` method with:
	* The `self` parameter
	* The additional parameters `make, model, year`
* `__init__()` once again takes in the parameters and assigns them to corresponding attributes
* These attributes will be associated with instances made from this class
* When a new car is made `make, model, year` all need to be specified
* In addition a method `get_descriptive_name()` is defined
* It puts a car's year, make, and model in a nice format
	* To work with the attributes the `self.` prefix is needed within the method
* As the class is right now it is relatively static, it can be expanded to include changing attributes

```python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer = 0  # default value — every new car starts at 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        return f"{self.year} {self.make} {self.model}".title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer} miles on it.")

new_car = Car('audi', 'a4', 2024)
print(new_car.get_descriptive_name())
new_car.read_odometer()
```

# Setting Default Attribute Values
---
* Attributes can be defined in `__init__()` **without** being passed as parameters
* Just assign a default value directly
* `odometer = 0` makes sense — every new car starts at 0 miles

* When an instance is created attributes can be defined without being passed as parameters
* These attributes can be defined in the `__init__()` method
	* They would have a default value assigned within
* Suppose we want to expand car to also have an `odometer_reading` that always starts with `0`
* When Python calls the `__init__()` method to create a new instance, it stores the make, model, and year
* But, Python also creates a new attribute `odometer` that it sets to 0
* The odometer will not always be `0` so we need a way to change it

# Modifying Attributes Directly
---
* The simplest way to modify an attribute is to access it **directly** using dot notation
* Use dot notation and assignment — same syntax as setting a variable

* The simplest way to modify the value of an attribute is to access the attribute directly
* This can be done using the dot operator and doing an assignment
* Dot notation is used to access the car's `odometer` attribute
* It then is able to directly set the value
* This tells python to:
	* Take the instance
	* Find the attribute `odometer` associated with it
	* Set the value of that attribute to be 13

```python
new_car = Car('audi', 'a4', 2024)
new_car.read_odometer()

# Modify the attribute directly
new_car.odometer = 13
new_car.read_odometer()
```

# Modifying Attributes Through a Method
---
* It can be helpful to have methods **update** certain attributes
* Pass the new value to a method — the method handles the change
* This lets you add **validation logic** — for example, preventing the odometer from going backwards

* It can be helpful to have methods update certain attributes
* Instead of accessing the attribute directly, you can pass the new value to a method
* The method then directly handles the changing of the variable
* The addition of `update_odometer()` takes in a mileage value and sets the odometer to it
* It also ensures we are only incrementing the odometer by positive values

```python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer = 0

    def get_descriptive_name(self):
        return f"{self.year} {self.make} {self.model}".title()

    def read_odometer(self):
        print(f"This car has {self.odometer} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading — reject rollbacks."""
        if mileage >= self.odometer:
            self.odometer = mileage
        else:
            print("You can't roll back an odometer!")

new_car = Car('audi', 'a4', 2024)
new_car.update_odometer(13)
new_car.read_odometer()

# Try to roll it back
new_car.update_odometer(5)
```

# Inheritance
---
* When one class **inherits** from another it takes on all its attributes and methods
* The original class is the **parent** (or superclass)
* The new class is the **child** (or subclass)
* The child can define new attributes and methods, or **override** parent ones

* You don't always have to start from scratch when writing a class
* If the class you're writing is a specialized version of another class you wrote, you can use *inheritance*
* When one class *inherits* from another it takes on the attributes and methods from the first class
* The original class is the *parent* class and the new class is the *child*
* The *child* can inherit any and all the attributes and methods from its parent class, or define new ones

# Writing a Child Class
---
* Include the **parent class name** in parentheses in the child class definition
* Call `super().__init__()` to initialize parent attributes
* `super()` lets Python call a method from the parent class
* The child automatically has access to all parent methods!

* When writing a new class based on an existing one, often the parent `__init__()` needs to be called
* This will initialize any attributes that were defined in the parent
* It also makes these attributes available to the child class
* An example could be modeling an electric car, which really is just a specific type of car
* When using inheritance the parent class must be in the current file
* Then a child class is defined, `ElectricCar`
* The name of the parent must be included in parenthesis in the definition of the child class
* The `super()` function is a special function that lets Python call a method from the parent class
* Specifically `super().__init__(make,model,year)` tells it to call the `__init__()` from the parent
* This gives an `ElectricCar` instance all the attributes defined in the parent `__init__()`
* The *super* name comes from a convention of calling the parent class a *superclass* and the child class a *subclass*

```python
class ElectricCar(Car):  # Car is the parent class
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """Initialize attributes of the parent class."""
        super().__init__(make, model, year)  # call parent __init__()

# ElectricCar automatically has all Car's methods!
leaf = ElectricCar('nissan', 'leaf', 2023)
print(leaf.get_descriptive_name())  # method from parent class
leaf.read_odometer()                 # method from parent class
```

# Defining Attributes and Methods for the Child
---
* Once the child inherits from the parent, add attributes and methods unique to the child
* These **won't** be available on instances of the parent class
* There's no limit to how much you can specialize a child class

* Once you have a child class defined that inherits from a parent class, new attributes and methods can be added
* So, for the electric cars concept we can add something unique to electric cars — how about battery size?
* A new attribute `battery_size` is added, with an initial value of 40
* This attribute will only be associated with instances created from `ElectricCar`
* It won't be associated with any instances of `Car`
* We also added a method just for the subclass
* An attribute that could belong to any kind of car, not just an electric one, should go in the parent class
* Then, anyone using `Car` will have that functionality available
* And the `ElectricCar` will only contain things for electric cars

```python
class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery_size = 40  # unique to electric cars

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")

leaf = ElectricCar('nissan', 'leaf', 2023)
print(leaf.get_descriptive_name())  # from parent
leaf.describe_battery()              # unique to child
```

# Overriding Parent Methods
---
* You can **override** any parent method that doesn't fit the child
* Define a method in the child with the **same name** as one in the parent
* Python will use the child's version and ignore the parent's

* You can override any method from the parent class that doesn't fit the child class
* To do this, define a method in the child class with the same name as one in the parent class
* Python will then disregard the parent class method and only pay attention to the child's method
* When using inheritance make child classes retain what they need
* Overwrite what they don't

```python
class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery_size = 40

    def describe_battery(self):
        print(f"This car has a {self.battery_size}-kWh battery.")

    def read_odometer(self):
        """Override parent method — electric cars track differently."""
        print(f"This electric car has {self.odometer} miles on it.")
        print("(Electric cars measure efficiency in MPGe, not MPG.)")

leaf = ElectricCar('nissan', 'leaf', 2023)
leaf.read_odometer()  # uses the child's version, not the parent's
```
