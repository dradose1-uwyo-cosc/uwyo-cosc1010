# Instances as Attributes (Composition)
---
* When a class gets large, part of it can be written as a **separate class**
* Large classes broken into smaller classes that work together — this is **composition**
* An instance of one class can be **stored as an attribute** of another

* For example we could add more details to the `ElectricCar` class outlining the battery
* Or, we could instead make a separate `Battery` class that deals only with the battery
* Then, an instance of battery can be used within the `ElectricCar` class
* There a new class `Battery` was defined, which doesn't inherit from any other class
* The `__init__()` method has one parameter, `battery_size`, in addition to `self`
* This is an optional parameter, as it has a default value associated with it
* `leaf.battery.describe_battery()` looks at the instance leaf, then finds the attribute `battery` within, which is another class instance
* While this may look like extra work, it ultimately allows `ElectricCar` to remain (relatively) uncluttered

```python
class Car:
    """A simple attempt to represent a car."""
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer = 0
    def get_descriptive_name(self):
        return f"{self.year} {self.make} {self.model}".title()


class Battery:
    """A simple attempt to model a battery for an electric car."""
    def __init__(self, battery_size=40):
        """Initialize the battery's attributes."""
        self.battery_size = battery_size

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")


class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery = Battery()  # Battery instance stored as an attribute


leaf = ElectricCar('nissan', 'leaf', 2023)
print(leaf.get_descriptive_name())
leaf.battery.describe_battery()  # access the Battery instance inside ElectricCar
```

# Importing Classes
---
* As files get long, classes can be stored in **separate modules**
* Import them into your main program — same pattern as importing functions
* Several import styles available:
  * `import module_name` — use `module_name.ClassName`
  * `from module_name import ClassName` — use `ClassName` directly
  * `from module_name import Class1, Class2` — import multiple classes
  * `import module_name as mn` — alias for brevity

* As you add more functionality to your classes, your files can get long
* Even if you use inheritance and composition well, you'll have a lot of code in one file
* Python lets you store classes in modules and import as needed
* This is the same pattern we saw with functions last week
* Connect back to functions importing from modules — the pattern is the same

```python
# If Car lived in car.py, you could import it like:

# Import the whole module
# import car
# my_car = car.Car('audi', 'a4', 2024)

# Import a specific class
# from car import Car
# my_car = Car('audi', 'a4', 2024)

# Import multiple classes
# from car import Car, ElectricCar

# Give module an alias
# import car as c
# my_car = c.Car('audi', 'a4', 2024)

# We already do this! datetime is a great real-world example:
from datetime import datetime
now = datetime.now()
print(f"Current date and time: {now}")
```

# The Python Standard Library
---
* A set of **modules included with every Python installation**
* Written by other programmers — you can use them to expand your code's functionality
* Import with an `import` statement
* The `random` module is a great example — `randint()` and `choice()` are immediately useful

* The *Python Standard Library* is a set of modules included with every Python installation
* These modules are things other programmers have written, and you can use them to expand your code's functionality
* You can use any function or class in the standard library by including an `import` statement
* To begin we will look at the `random` module
* It includes a helpful function `randint()` among others
* This function takes two integers and returns a randomly selected integer between those numbers, inclusive
* `random` also allows you to randomly pick an element out of a list or tuple with `choice()`
* `random` shouldn't be used for security related projects

```python
from random import randint, choice

# Random integer between two values, inclusive
print(randint(1, 6))   # like rolling a die
print(randint(1, 6))

# Randomly pick from a list
cities = ['Laramie', 'Casper', 'Cheyenne', 'Sheridan', 'Gillette']
print(choice(cities))
```

# Styling Classes
---
* Class names use **CamelCase** — capitalize the first letter of every word, no underscores
* Instance and module names use **lowercase with underscores**
* Every class needs a **docstring** immediately after the class definition
* All `import` statements at the **top of the file**
* Two blank lines between classes

* There are a few styling issues related to classes
* Class names should be written in *CamelCase*
	* The first letter of every word should be capitalized
	* Underscores aren't used
* Instance and module names should be written in lowercase, with underscores between words
* Every class should have a docstring immediately following the class definition
* The docstring should be a brief description of what the class does
* The same formatting conventions should be followed as writing function docstrings
* Blank lines can be used to organize code, but not excessively
* When importing things the imports should occur at the top of the file

```python
# Good class styling example

class ElectricCar(Car):          # CamelCase class name
    """Represent an electric vehicle."""

    def __init__(self, make, model, year):
        """Initialize parent attributes and EV-specific attributes."""
        super().__init__(make, model, year)
        self.battery = Battery()  # instance name is lowercase


my_leaf = ElectricCar('nissan', 'leaf', 2023)  # instance name lowercase
print(my_leaf.get_descriptive_name())
```
